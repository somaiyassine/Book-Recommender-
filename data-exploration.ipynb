# Step 1 — Data exploration & cleaning

**Goal:** download the raw books dataset, understand its quality, and keep only the books that are useful for a recommender.

**Output:** `books_cleaned.csv` — the input of every following notebook.

## 1.1 Download the dataset
We use [`kagglehub`](https://github.com/Kaggle/kagglehub) to download the *7k Books with Metadata* dataset from Kaggle. The files are cached locally, so the download only happens once.

```python
import kagglehub
import seaborn
from seaborn import heatmap

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

## 1.2 Load the data
Read the CSV into a pandas DataFrame and take a first look at the columns (title, authors, categories, description, rating, ...).

```python
import pandas as pd
```

```python
books = pd.read_csv(f"{path}/books.csv")
```

```python
books
```

## 1.3 Where is data missing?
A heatmap of `isna()` shows the missing values: each light line is a missing value. This tells us which columns we can rely on.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(),cbar=False,ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing Values")
plt.show()
```

## 1.4 Is "missing description" biased?
Before dropping books without a description, we check that those books are not special (e.g. all very old or very short).
We create two helper columns:
- `missing_description` — 1 if the description is missing, else 0
- `age_of_book` — how old the book is

```python
import numpy as np
books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2026 - books["published_year"]
```

Spearman correlation between the helper columns. Values close to 0 mean **no strong relationship**, so removing books with a missing description will not bias the dataset.

```python
columns_of_interest = ["num_pages", "age_of_book", "missing_description", "average_rating"]
correlation_matrix = books[columns_of_interest].corr(method="spearman")
sns.set_theme(style="white")
plt.figure(figsize=(10,6))
heatmap= sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar_kws={"label": "Spearman"})
heatmap.set_title("Correlation heatmap")
plt.show()
```

## 1.5 Remove incomplete books
Keep only books that have a description, a number of pages, an average rating and a publication year.

```python
book_missing = books[~(books["description"].isna()) &
      ~(books["num_pages"].isna()) &
      ~(books["average_rating"].isna()) &
      ~(books["published_year"].isna())
]
book_missing
```

The `categories` column has hundreds of different values, most with very few books — we will simplify them later in the text-classification step.

```python
book_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
book_missing
```

## 1.6 How long are the descriptions?
Count the words in each description. Very short descriptions carry too little meaning for semantic search.

```python
book_missing["words_in_description"] = book_missing["description"].str.split().str.len()
```

```python
book_missing
```

Look at descriptions in different length ranges to decide on a cut-off. Under ~25 words they are mostly useless (e.g. *"A novel"*, *"Donation"*).

```python
book_missing.loc[book_missing["words_in_description"].between(1, 4), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(5, 14), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(15, 24), "description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(25, 34), "description"]
```

## 1.7 Keep descriptions with at least 25 words

```python
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25]
book_missing_25_words
```

## 1.8 Build helper columns
- `title_and_subtitle` — title and subtitle joined with `": "` (many titles only make sense with their subtitle).

```python
book_missing_25_words["title_and_subtitle"] = (
     book_missing_25_words["title"].fillna("").astype("string")
     + book_missing_25_words["subtitle"].fillna("").astype("string").radd(": ")
         .where(book_missing_25_words["subtitle"].notna(), "")
 )
book_missing_25_words
```

- `tagged_description` — the ISBN followed by the description, e.g. `9780002005883 A NOVEL THAT READERS...`.
  The ISBN "tag" lets us find the book again after a vector search returns only the text.

```python
book_missing_25_words["tagged_description"] = book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
```

```python
book_missing_25_words
```

## 1.9 Save the cleaned dataset
Drop the temporary helper columns and save to `books_cleaned.csv`.

```python
(
    book_missing_25_words
    .drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
    .to_csv("books_cleaned.csv", index = False)
)
```
